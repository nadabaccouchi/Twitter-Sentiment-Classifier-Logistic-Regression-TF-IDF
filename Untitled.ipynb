```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

```python
def clean_text(text):
    if isinstance(text, str):
        text = text.lower()
        text = re.sub(r"http\S+|www\S+|https\S+", '', text)   # remove URLs
        text = re.sub(r'@\w+|#', '', text)                    # remove @mentions & hashtags
        text = re.sub(r'[^a-z\s]', '', text)                  # keep letters only
        words = [lemmatizer.lemmatize(word) 
                 for word in text.split() if word not in stop_words]
        return " ".join(words)
    return ""
```

```python
data['Clean_Content'] = data['Content'].apply(clean_text)
```

```python
# ---------- Custom cleaning that preserves negation and pronouns ----------
import re
import string

# If NLTK is available, use it; otherwise use a small fallback list.
try:
    import nltk
    from nltk.corpus import stopwords
    from nltk.stem import WordNetLemmatizer
    nltk.data.find('corpora/stopwords')
    nltk.data.find('corpora/wordnet')
    lemmatizer = WordNetLemmatizer()
    base_stop = set(stopwords.words('english'))
except Exception:
    # offline fallback (small)
    lemmatizer = None
    base_stop = {
        'the','a','an','and','is','in','at','of','for','to','with','as','on','by','that','this',
        'are','be','was','were','it','from','or','but','if','they','their','them'
    }

# Words we DO NOT want removed because they affect meaning
keep_words = {
    'not','no','never','don','dont',"don't","doesn","doesn't","didn","didn't",
    'i','you','we','they','he','she','all','will','would','should','could','won'
}

custom_stop = base_stop - keep_words

# simple contraction expansion (extend as needed)
contractions = {
    "don't": "do not", "doesn't": "does not", "didn't": "did not",
    "won't": "will not", "can't": "can not", "couldn't": "could not",
    "i'm": "i am", "i'll": "i will", "you're": "you are", "you'll": "you will",
    "we're": "we are", "they're": "they are", "i've": "i have", "i'd": "i would"
}

def expand_contractions(text):
    for k, v in contractions.items():
        text = re.sub(r"\b" + re.escape(k) + r"\b", v, text, flags=re.IGNORECASE)
    return text

def clean_text_preserve(text,
                        keep_pronouns=True,
                        keep_negations=True,
                        lemmatize_if_possible=True):
    """
    Clean text but preserve negations and pronouns (configurable).
    - keep_pronouns: if True, pronouns like 'i', 'you' are NOT removed.
    - keep_negations: if True, words like 'not','no','never' are NOT removed.
    - lemmatize_if_possible: will lemmatize if nltk.wordnet is available.
    """
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = expand_contractions(text)

    # remove urls and mentions, leave normal words & spaces
    text = re.sub(r"http\S+|www\S+|https\S+", '', text)
    text = re.sub(r'@\w+|#', '', text)                 # remove @mentions and hashtag signs
    text = re.sub(r"[^a-z\s']", " ", text)            # keep letters and apostrophes (mostly expanded)

    # choose stopword set based on flags
    stop_set = set(custom_stop)  # custom_stop built earlier
    if not keep_pronouns:
        stop_set |= {'i','you','we','they','he','she','me','him','her','us','them'}
    if not keep_negations:
        stop_set |= {'not','no','never','don','dont','won','cannot','can','will','would','should','could'}

    tokens = [w for w in text.split() if w and w not in stop_set]

    if lemmatize_if_possible and lemmatizer is not None:
        tokens = [lemmatizer.lemmatize(w) for w in tokens]

    return " ".join(tokens)

# ---------------- Example - run these after defining the function ----------------
example = "im getting on borderlands and i will murder you all ,"
print("ORIGINAL:   ", example)
print("CLEAN (default preserve neg/pron):", clean_text_preserve(example))
print("CLEAN (remove pronouns but keep negations):", clean_text_preserve(example, keep_pronouns=False))
print("CLEAN (remove pron & neg):", clean_text_preserve(example, keep_pronouns=False, keep_negations=False))
```
